```python
# 準備一個初始化的清單
# 1. 如果初始化都依樣
l = [0, 1] * 5
print(l)
# 2. 如果初始化值不一樣
l = []
for i in range(5):
    l.append(i)
print(l)
# 3. 2的簡化
l = [(i+1) for i in range(5)]
print(l)
```

```python
# 雙層清單的初始化
# 錯誤方法: 裡面的清單只創造了一次, 永遠都是那份
a = [0] * 3
l = [a] * 5
print(l)
l[1][0] = 99
print(l)
# 正確方法
# 2.
l = []
for i in range(5):
    a = [0] * 3
    l.append(a)
l[1][0] = 99
print(l)
# 3.
l = [[0] * 3 for i in range(5)]
l[1][0] = 88
print(l)
```
