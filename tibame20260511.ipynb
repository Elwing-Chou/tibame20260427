```python
# 準備一個初始化的清單
# 1. 如果初始化都依樣
l = [0, 1] * 5
print(l)
# 2. 如果初始化值不一樣
l = []
for i in range(5):
    l.append(i)
print(l)
# 3. 2的簡化
l = [(i+1) for i in range(5)]
print(l)
```

```python
# 雙層清單的初始化
# 錯誤方法: 裡面的清單只創造了一次, 永遠都是那份
a = [0] * 3
l = [a] * 5
print(l)
l[1][0] = 99
print(l)
# 正確方法
# 2.
l = []
for i in range(5):
    a = [0] * 3
    l.append(a)
l[1][0] = 99
print(l)
# 3.
l = [[0] * 3 for i in range(5)]
l[1][0] = 88
print(l)
```

```python
# [
#   [0, 1, 2, 3, 4],
#   [9, 8, 7, 6, 5],
#   [10, 11, 12, 13, 14]
#   [19, 18, 17, 16, 15]
# ]
row, col = 4, 5
result = [[0] * col for i in range(row)]

start = 8
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            result[i][j] = col * i + j + start
        else:
            result[i][j] = (col * i + (col - 1)) - j + start
print(result)
```

Python群集
1. list: 放同類型的東西 [20, 30, 50], key(python幫你設定的)=0,1,2,3 -> 有序集合
2. dict: 聚集欄位形成一個複雜單位(自定義的key:value) {"姓名":"Elwing", "身高":175} -> 無序集合
3. set: 決不重複的集合 {30, 40, 20}
沒有key讓你使用 -> 無序集合
4. tuple: 簡化版本的字典 {"r":255, "g":128, "b":127} -> (255, 128, 127) key=0,1,2 -> 有序集合

```python
s = {40, 30, 20, 20}
print(s)
# 交集(&): 我們都有的才算
# 聯集(|): 把我們全部的並列
# 減集(-): 我 - 我們兩個重複的(&)
a = {20, 30}
b = {30, 40}
print(a & b)
print(a | b)
print(a - b)
# 專屬功能: add(很類似list append)
a.add(88)
print(a)
# for...in
for n in a | b:
    print("!", n)
```

```python
# 樂透號碼(6個不重複的號碼)
import random
lot = []
while True:
    n = random.randint(1, 49)
    print(n)
    if not n in lot:
        lot.append(n)
    if len(lot) == 6:
        break
print(lot)
```

```python
import random
lot = set()
while True:
    n = random.randint(1, 49)
    print(n)
    lot.add(n)
    if len(lot) == 6:
        break
print(lot)
```

```python
import pygame as pg
import random

#pygame初始化
pg.init()

# --- 字型 ---
def get_font(size):
    for f in ['microsoftjhenghei', 'simhei', 'stheitirelight']:
        if f in pg.font.get_fonts():
            return pg.font.SysFont(f, size)
    return pg.font.SysFont(None, size)


# GAME UI
FONT_UI = get_font(32)
col, row = 15, 15
inter = 40
width, height = (col + 2) * inter, (row + 2) * inter
# 產生視窗
screen = pg.display.set_mode([width, height])
# 設定遊戲標題
pg.display.set_caption("踩地雷")

# 準備第一個圖層
bg = pg.Surface(screen.get_size())
# 把畫布填滿某個顏色
bg.fill([199, 167, 82])

# 2. 畫地雷
# pygame.draw.rect(畫布, 顏色, [x坐標, y坐標, 寬度, 高度], 線寬)
pg.draw.rect(bg, [255, 0, 0], [80, 80, inter, inter], 0)
# 3. 畫數字
pg.draw.rect(bg, [255, 255, 255], [120, 80, inter, inter], 0)
# render文字
t = FONT_UI.render("4", 1, [0, 0, 0])
# 背景(bg)上面疊上t get_rect(中心座標) -> 左上角座標
bg.blit(t, t.get_rect(center=[140, 100]))

# 1. 畫棋盤
# pygame.draw.line(畫布, 顏色, (x坐標1, y坐標1), (x坐標2, y坐標2), 線寬)
for i in range(row+1):
    pg.draw.line(bg,
                 [0, 0, 0],
                 [inter, inter*i+inter],
                 [width-inter, inter*i+inter],
                 1)

for i in range(col+1):
    pg.draw.line(bg,
                 [0, 0, 0],
                 [inter*i+inter, inter],
                 [inter*i+inter, height-inter],
                 1)

# 你要把圖層放到上一層
screen.blit(bg, [0, 0])
# 對畫面進行更新(才會真的秀出來)
pg.display.update()


# 建立一個永不結束的迴圈(遊戲才不會結束)
running = True
while running:
    # 收取你的遊戲任何事件(滑鼠點擊/鍵盤按鈕...)
    for event in pg.event.get():
        # 偵測滑鼠點擊以後放掉的動作
        if event.type == pg.MOUSEBUTTONUP:
            pass
        # 如果收到的事件是按x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False

pg.quit()
```
