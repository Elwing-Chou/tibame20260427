```python
# 準備一個初始化的清單
# 1. 如果初始化都依樣
l = [0, 1] * 5
print(l)
# 2. 如果初始化值不一樣
l = []
for i in range(5):
    l.append(i)
print(l)
# 3. 2的簡化
l = [(i+1) for i in range(5)]
print(l)
```
