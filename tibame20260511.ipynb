```python
# 準備一個初始化的清單
# 1. 如果初始化都依樣
l = [0, 1] * 5
print(l)
# 2. 如果初始化值不一樣
l = []
for i in range(5):
    l.append(i)
print(l)
# 3. 2的簡化
l = [(i+1) for i in range(5)]
print(l)
```

```python
# 雙層清單的初始化
# 錯誤方法: 裡面的清單只創造了一次, 永遠都是那份
a = [0] * 3
l = [a] * 5
print(l)
l[1][0] = 99
print(l)
# 正確方法
# 2.
l = []
for i in range(5):
    a = [0] * 3
    l.append(a)
l[1][0] = 99
print(l)
# 3.
l = [[0] * 3 for i in range(5)]
l[1][0] = 88
print(l)
```

```python
# [
#   [0, 1, 2, 3, 4],
#   [9, 8, 7, 6, 5],
#   [10, 11, 12, 13, 14]
#   [19, 18, 17, 16, 15]
# ]
row, col = 4, 5
result = [[0] * col for i in range(row)]

start = 8
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            result[i][j] = col * i + j + start
        else:
            result[i][j] = (col * i + (col - 1)) - j + start
print(result)
```

```python
s = {40, 30, 20, 20}
print(s)
# 交集(&): 我們都有的才算
# 聯集(|): 把我們全部的並列
# 減集(-): 我 - 我們兩個重複的(&)
a = {20, 30}
b = {30, 40}
print(a & b)
print(a | b)
print(a - b)
# 專屬功能: add(很類似list append)
a.add(88)
print(a)
# for...in
for n in a | b:
    print("!", n)
```

```python
# 樂透號碼(6個不重複的號碼)
import random
lot = []
while True:
    n = random.randint(1, 49)
    print(n)
    if not n in lot:
        lot.append(n)
    if len(lot) == 6:
        break
print(lot)
```

```python
import random
lot = set()
while True:
    n = random.randint(1, 49)
    print(n)
    lot.add(n)
    if len(lot) == 6:
        break
print(lot)
```
