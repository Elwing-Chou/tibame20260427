````python
import urllib.request as req
import bs4 as bs

url = "https://tabelog.com/tw/tokyo/rstLst/sweets/?SrtT=rt"
resp = req.urlopen(url)
content = resp.read()
html = bs.BeautifulSoup(content)
html.find("a", {"class":"list-rst__rst-name-target"})
````



```

class="分類1 分類2"
2. 透過class做選擇

}

.js-thumbnail-img{
```
